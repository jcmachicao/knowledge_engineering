```python
class SwarmCognitiveSystem:
    """
    Sistema cognitivo sin controlador central
    Inspirado en: hormigas, abejas, peces, aves
    """

    def __init__(self, n_agents=100):
        self.agents = [SimpleAgent(id=i) for i in range(n_agents)]
        self.environment = SharedEnvironment()
        self.global_behavior = None  # Emerge, no se programa

    def simulate_ant_colony_optimization(self, problem):
        """
        Hormigas resolviendo problemas complejos
        SIN LÍDER, SIN PLAN CENTRAL
        """
        # Cada hormiga:
        # 1. Explora localmente
        # 2. Deja feromonas en buenos caminos
        # 3. Sigue feromonas de otras

        for iteration in range(max_iterations):
            for ant in self.agents:
                # Decisión local basada en:
                # - Feromonas locales (memoria distribuida)
                # - Exploración aleatoria
                path = ant.explore_locally(self.environment)

                if path.is_solution():
                    ant.deposit_pheromone(path, strength=1/path.length)

            # Las feromonas se evaporan (olvido distribuido)
            self.environment.evaporate_pheromones(rate=0.1)

        # La SOLUCIÓN EMERGE del comportamiento colectivo
        best_path = self.environment.get_strongest_pheromone_trail()
        return best_path
```

###1.3 Sistema Inmune como Sistema Cognitivo

```python
class ImmuneCognitiveSystem:
    """
    El sistema inmune como sistema de aprendizaje distribuido
    Reconocimiento de patrones sin cerebro
    """

    def __init__(self):
        self.b_cells = Population(size='1e9')  # Células B
        self.t_cells = Population(size='1e9')  # Células T
        self.memory_cells = []  # Memoria inmunológica

    def recognize_pathogen(self, pathogen):
        """
        Reconocimiento de patrones sin procesamiento central
        """
        # 1. DIVERSIDAD: Generar variaciones aleatorias
        for cell in self.b_cells:
            cell.antibody = random_variation()

        # 2. SELECCIÓN: Células que "reconocen" patógeno se multiplican
        matching_cells = [
            cell for cell in self.b_cells
            if cell.antibody.binds_to(pathogen)
        ]

        # 3. AMPLIFICACIÓN: Selección clonal
        for cell in matching_cells:
            cell.proliferate(copies=1000)

        # 4. MEMORIA: Algunas células se vuelven memoria
        self.memory_cells.extend(
            cell.to_memory_cell() for cell in matching_cells
        )

        # 5. RESPUESTA más rápida en futuros encuentros
        # = APRENDIZAJE DISTRIBUIDO
```

#🌐 3. COGNICIÓN EXTENDIDA (EXTENDED COGNITION)
###3.1 La Mente Extendida (Extended Mind Thesis)

```python
class ExtendedMind:
    """
    La cognición no está limitada al cerebro
    Se extiende a herramientas y ambiente
    Clark & Chalmers (1998)
    """

    def otto_and_inga_example(self):
        """
        Experimento mental clásico sobre memoria extendida
        """

        class Inga:
            def __init__(self):
                self.biological_memory = Memory()

            def remember_address(self):
                # Inga recuerda dirección del museo
                return self.biological_memory.retrieve("MOMA_address")

        class Otto:
            def __init__(self):
                self.biological_memory = DementiaAffectedMemory()
                self.notebook = Notebook()  # Memoria externa

            def remember_address(self):
                # Otto tiene Alzheimer, usa libreta
                return self.notebook.lookup("MOMA_address")

        # PREGUNTA: ¿Es la libreta de Otto parte de su "mente"?
        # RESPUESTA: Sí, funcionalmente equivalente a memoria biológica

        # CRITERIOS para cognición extendida:
        # ✓ Constante disponibilidad (Otto siempre lleva libreta)
        # ✓ Confiabilidad (Otto confía en libreta como Inga en su memoria)
        # ✓ Fácil acceso (consultar libreta es automático)
        # ✓ Información respaldada previamente (Otto escribió notas)

# IMPLICACIONES PARA IA:
class ExtendedAISystem:
    def __init__(self):
        self.llm = LanguageModel()  # "Cerebro"
        self.vector_db = VectorDatabase()  # "Memoria extendida"
        self.code_interpreter = PythonREPL()  # "Herramienta cognitiva"
        self.web_browser = Browser()  # "Acceso a conocimiento externo"

    def solve_problem(self, task):
        # El sistema cognitivo incluye:
        # - LLM (procesamiento)
        # - Vector DB (memoria de largo plazo)
        # - Code interpreter (razonamiento formal)
        # - Web (conocimiento colectivo)

        # La "mente" del sistema está distribuida
        # No es solo el LLM

        context = self.vector_db.retrieve_relevant(task)  # Memoria
        code = self.llm.generate_code(task, context)  # Razonamiento
        result = self.code_interpreter.execute(code)  # Verificación
        external_info = self.web_browser.search(task)  # Conocimiento externo

        # Cognición = LLM + Tools + External Resources
        return self.llm.synthesize(result, external_info)
```

###3.2 Cognición Socialmente Distribuida

```python
class DistributedCognition:
    """
    Cognición distribuida entre múltiples agentes
    Hutchins (1995) - "Cognition in the Wild"
    """

    def airplane_cockpit_cognition(self):
        """
        Caso clásico: Navegación aérea como sistema cognitivo distribuido
        """

        class Cockpit:
            def __init__(self):
                self.pilot = Pilot()
                self.copilot = Copilot()
                self.instruments = Instruments()
                self.checklist = Checklist()  # Memoria externa

            def navigate_approach(self, airport):
                # COGNICIÓN DISTRIBUIDA:

                # 1. Memoria distribuida
                knowledge = {
                    'pilot': self.pilot.experience,
                    'copilot': self.copilot.training,
                    'checklist': self.checklist.procedures,
                    'instruments': self.instruments.state
                }

                # 2. Procesamiento distribuido
                self.pilot.assess_situation()
                self.copilot.monitor_instruments()
                self.instruments.display_critical_info()

                # 3. Decisión emerge de interacción
                decision = self.pilot.command + \
                          self.copilot.verification + \
                          self.instruments.feedback

                # El SISTEMA COMPLETO es la unidad cognitiva
                # No el piloto individual

                return decision

        # LECCIONES:
        # - Cognición es propiedad del SISTEMA, no individuos
        # - Representaciones están distribuidas (mente + papel + pantallas)
        # - Transformación de representaciones es el "cómputo"

# APLICACIÓN A SISTEMAS LLM MULTI-AGENTE:
class MultiAgentCognitiveSystem:
    def __init__(self):
        self.agents = {
            'researcher': LLM(role="research"),
            'critic': LLM(role="critique"),
            'synthesizer': LLM(role="synthesis"),
            'validator': LLM(role="validation")
        }
        self.shared_workspace = SharedMemory()

    def collaborative_problem_solving(self, problem):
        # Cognición distribuida entre agentes LLM

        # Fase 1: Investigación (agente researcher)
        findings = self.agents['researcher'].investigate(problem)
        self.shared_workspace.add(findings)

        # Fase 2: Crítica (agente critic)
        critique = self.agents['critic'].analyze(findings)
        self.shared_workspace.add(critique)

        # Fase 3: Síntesis (agente synthesizer)
        synthesis = self.agents['synthesizer'].integrate(
            findings, critique
        )

        # Fase 4: Validación (agente validator)
        validated = self.agents['validator'].verify(synthesis)

        # La solución emerge de interacción distribuida
        # NO de un solo LLM

        return validated
```

```python

class NonNeuralIntelligence:
    """
    Inteligencia en organismos sin sistema nervioso
    """

    def slime_mold_maze_solving(self):
        """
        Physarum polycephalum resuelve laberintos óptimamente
        SIN NEURONAS
        """

        class SlimeMold:
            def __init__(self):
                self.body = ProtoplasmaticNetwork()
                self.neurons = 0  # Sin sistema nervioso

            def solve_maze(self, maze, food_sources):
                # 1. El moho crece llenando todo el laberinto
                self.body.expand_through(maze)

                # 2. Retrae tubos que NO conectan fuentes de comida
                for tube in self.body.tubes:
                    if not tube.connects_food_sources():
                        tube.retract()

                # 3. Fortalece tubos que SÍ conectan comida
                for tube in self.body.tubes:
                    if tube.connects_food_sources():
                        tube.reinforce()

                # RESULTADO: Camino óptimo emerge
                # Aproxima algoritmo de Steiner tree
                # TODO mediante procesos bioquímicos

                return self.body.optimal_network()

        # APLICACIONES:
        # - Diseño de redes de transporte
        # - Tokyo utilizó esto para optimizar sistema de trenes

    def plant_cognition(self):
        """
        Plantas muestran comportamientos "cognitivos"
        """

        class Plant:
            def __init__(self):
                self.roots = RootSystem()
                self.neurons = 0

            def root_foraging(self):
                """
                Raíces "buscan" nutrientes
                Muestran memoria y aprendizaje
                """
                # MEMORIA: Raíces "recuerdan" dónde había nutrientes
                if self.roots.previously_found_nutrients_here():
                    self.roots.grow_more_in_direction()

                # APRENDIZAJE: Distinguen señales confiables
                if self.roots.reliable_signal_detected():
                    self.roots.respond_stronger()
                else:
                    self.roots.ignore_signal()

                # COMUNICACIÓN: Red de hongos (Wood Wide Web)
                self.roots.send_signal_through_mycelium(
                    message="nutrients_here"
                )

            def associative_learning(self):
                """
                Plantas aprenden asociaciones (Pavlov vegetal)
                Gagliano et al. (2016)
                """
                # Experimento: Luz + Ventilador → Luz predice ventilador

                for trial in range(training_trials):
                    self.experience(light=True)
                    time.sleep(1)
                    self.experience(wind=True)

                # Después de entrenamiento:
                self.experience(light=True)
                # → Planta se prepara para viento (inclina hojas)
                # = APRENDIZAJE ASOCIATIVO sin cerebro
```
