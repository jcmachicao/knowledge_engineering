# Sistemas Expertos: Introducción Conceptual y Práctica

## 1. Introducción

Los sistemas expertos representan una de las primeras aplicaciones exitosas de la Inteligencia Artificial, surgidos en las décadas de 1960-1970. Fueron diseñados para emular la toma de decisiones de un experto humano en dominios específicos y bien delimitados.

**Definición**: Un sistema experto es un programa computacional que utiliza conocimiento especializado y procedimientos de inferencia para resolver problemas que normalmente requieren experticia humana significativa.

## 2. Características Fundamentales

### 2.1 Elementos Distintivos

- **Especialización en dominio**: Operan en áreas específicas y acotadas (diagnóstico médico, configuración de sistemas, análisis químico, etc.)
- **Conocimiento explícito**: El conocimiento está codificado de forma declarativa y separada del mecanismo de inferencia
- **Razonamiento simbólico**: Manipulan símbolos y reglas, no realizan cálculos numéricos complejos
- **Explicación de razonamiento**: Capacidad de justificar sus conclusiones mostrando la cadena de inferencias
- **Manejo de incertidumbre**: Pueden trabajar con información incompleta o incierta

### 2.2 ¿Qué NO son los Sistemas Expertos?

Para evitar confusiones futuras con arquitecturas cognitivas:
- No son sistemas de propósito general
- No aprenden de forma autónoma (en su concepción clásica)
- No integran múltiples tipos de razonamiento
- No modelan la cognición humana completa
- No tienen memoria episódica o contextual compleja

## 3. Arquitectura Básica

Un sistema experto clásico consta de tres componentes principales:

### 3.1 Base de Conocimiento (Knowledge Base)

Almacena el conocimiento del dominio, típicamente en forma de:
- **Reglas de producción** (IF-THEN)
- **Hechos** (proposiciones verdaderas en el dominio)
- **Relaciones** entre entidades del dominio

Ejemplo conceptual:
```
SI (fiebre > 38.5) Y (tos persistente) Y (dificultad respiratoria)
ENTONCES (sospecha = neumonía) CON certeza 0.8
```

### 3.2 Motor de Inferencia (Inference Engine)

Mecanismo que aplica reglas lógicas al conocimiento para derivar conclusiones. Dos estrategias principales:

**Encadenamiento hacia adelante (Forward Chaining)**:
- Parte de los hechos conocidos
- Aplica reglas hasta alcanzar conclusiones
- Orientado a datos (data-driven)
- Útil para: diagnóstico, monitoreo, control

**Encadenamiento hacia atrás (Backward Chaining)**:
- Parte de una hipótesis o meta
- Busca hechos que la confirmen o refuten
- Orientado a objetivos (goal-driven)
- Útil para: planificación, consulta, verificación

### 3.3 Base de Hechos o Memoria de Trabajo (Working Memory)

Almacena:
- Hechos específicos del problema actual
- Resultados intermedios durante la inferencia
- Estado temporal de la sesión

### 3.4 Interfaz de Usuario

Permite:
- Entrada de hechos y consultas
- Visualización de conclusiones
- Explicación del razonamiento

## 4. Proceso de Construcción

### 4.1 Adquisición del Conocimiento

El cuello de botella tradicional de los sistemas expertos:
1. Identificación del experto humano
2. Extracción del conocimiento (mediante entrevistas, observación)
3. Formalización en reglas
4. Validación con casos reales
5. Refinamiento iterativo

### 4.2 Representación del Conocimiento

Formatos comunes:
- **Reglas de producción**: IF condición THEN acción
- **Marcos (frames)**: estructuras de datos con atributos
- **Redes semánticas**: grafos de conceptos y relaciones
- **Lógica de predicados**: expresiones formales lógicas

## 5. Ventajas y Limitaciones

### Ventajas
- Preservan conocimiento experto valioso
- Consistencia en las decisiones
- Disponibilidad 24/7
- Capacidad de explicación
- Facilitan la transferencia de conocimiento

### Limitaciones
- Conocimiento estático (no aprenden)
- Frágiles ante situaciones no previstas
- Mantenimiento costoso
- Dificultad para manejar sentido común
- Limitados a dominios estrechos

## 6. Ejemplos Históricos Relevantes

- **MYCIN (1970s)**: Diagnóstico de infecciones bacterianas
- **DENDRAL (1965)**: Análisis de estructuras químicas
- **XCON/R1 (1980s)**: Configuración de computadoras VAX
- **PROSPECTOR (1970s)**: Exploración geológica

### Ejercicio 1: Sistema Experto Simple con Reglas

Implementaremos un sistema experto básico para diagnóstico de problemas en computadoras.

````python
class SistemaExperto:
    def __init__(self):
        self.base_conocimiento = []
        self.hechos = set()

    def agregar_regla(self, condiciones, conclusion, certeza=1.0):
        """
        Agrega una regla a la base de conocimiento.
        condiciones: lista de hechos necesarios
        conclusion: hecho derivado
        certeza: factor de certeza (0 a 1)
        """
        self.base_conocimiento.append({
            'condiciones': condiciones,
            'conclusion': conclusion,
            'certeza': certeza
        })

    def agregar_hecho(self, hecho):
        """Agrega un hecho a la memoria de trabajo"""
        self.hechos.add(hecho)

    def encadenamiento_adelante(self):
        """
        Motor de inferencia con encadenamiento hacia adelante
        """
        hechos_nuevos = True
        iteracion = 0

        while hechos_nuevos:
            hechos_nuevos = False
            iteracion += 1
            print(f"\n--- Iteración {iteracion} ---")
            print(f"Hechos actuales: {self.hechos}")

            for regla in self.base_conocimiento:
                # Verificar si todas las condiciones se cumplen
                if all(cond in self.hechos for cond in regla['condiciones']):
                    # Si la conclusión no está ya en los hechos
                    if regla['conclusion'] not in self.hechos:
                        print(f"Aplicando regla: {regla['condiciones']} → {regla['conclusion']}")
                        self.hechos.add(regla['conclusion'])
                        hechos_nuevos = True

        print(f"\n=== Inferencia completa en {iteracion} iteraciones ===")
        return self.hechos
````

````python
# Crear el sistema experto
sistema = SistemaExperto()

# Definir reglas para diagnóstico de problemas de computadora
sistema.agregar_regla(
    ['computadora_no_enciende', 'luz_led_apagada'],
    'problema_alimentacion'
)

sistema.agregar_regla(
    ['problema_alimentacion', 'cable_conectado'],
    'fuente_poder_dañada'
)

sistema.agregar_regla(
    ['computadora_enciende', 'no_hay_imagen', 'beep_continuo'],
    'problema_ram'
)

sistema.agregar_regla(
    ['computadora_enciende', 'no_hay_imagen', 'sin_beep'],
    'problema_tarjeta_video'
)

sistema.agregar_regla(
    ['computadora_lenta', 'disco_lleno'],
    'necesita_limpieza_disco'
)

# Agregar hechos observados
print("=== SISTEMA EXPERTO: DIAGNÓSTICO DE COMPUTADORAS ===\n")
sistema.agregar_hecho('computadora_no_enciende')
sistema.agregar_hecho('luz_led_apagada')
sistema.agregar_hecho('cable_conectado')

# Ejecutar inferencia
conclusiones = sistema.encadenamiento_adelante()

print(f"\nConclusiones finales: {conclusiones}")
````

### Ejercicio 2: Sistema Experto con Factores de Certeza

Este ejercicio extiende el anterior para manejar incertidumbre.

````python
class SistemaExpertoCerteza:
    def __init__(self):
        self.reglas = []
        self.hechos = {}  # {hecho: certeza}

    def agregar_regla(self, condiciones, conclusion, certeza_regla=1.0):
        self.reglas.append({
            'condiciones': condiciones,
            'conclusion': conclusion,
            'certeza': certeza_regla
        })

    def agregar_hecho(self, hecho, certeza=1.0):
        self.hechos[hecho] = certeza

    def calcular_certeza_condiciones(self, condiciones):
        """
        Calcula la certeza mínima de un conjunto de condiciones
        """
        certezas = [self.hechos.get(cond, 0) for cond in condiciones]
        return min(certezas) if certezas else 0

    def inferir(self):
        cambios = True
        iteracion = 0

        while cambios:
            cambios = False
            iteracion += 1
            print(f"\n--- Iteración {iteracion} ---")

            for regla in self.reglas:
                # Verificar si todas las condiciones existen
                if all(cond in self.hechos for cond in regla['condiciones']):
                    # Calcular certeza de la conclusión
                    certeza_conds = self.calcular_certeza_condiciones(regla['condiciones'])
                    certeza_conclusion = certeza_conds * regla['certeza']

                    # Si la conclusión no existe o tiene menor certeza
                    if regla['conclusion'] not in self.hechos:
                        print(f"Nueva conclusión: {regla['conclusion']} "
                              f"(certeza: {certeza_conclusion:.2f})")
                        self.hechos[regla['conclusion']] = certeza_conclusion
                        cambios = True
                    elif certeza_conclusion > self.hechos[regla['conclusion']]:
                        print(f"Actualizando certeza de {regla['conclusion']}: "
                              f"{self.hechos[regla['conclusion']]:.2f} → {certeza_conclusion:.2f}")
                        self.hechos[regla['conclusion']] = certeza_conclusion
                        cambios = True

        return self.hechos
````

````python
# Crear sistema para diagnóstico médico simplificado
sistema = SistemaExpertoCerteza()

# Reglas con diferentes niveles de certeza
sistema.agregar_regla(['fiebre', 'tos'], 'gripe', 0.7)
sistema.agregar_regla(['fiebre', 'dolor_garganta'], 'gripe', 0.6)
sistema.agregar_regla(['fiebre', 'tos', 'dificultad_respirar'], 'neumonia', 0.8)
sistema.agregar_regla(['gripe', 'dificultad_respirar'], 'complicacion_respiratoria', 0.75)

# Hechos observados con certeza
print("=== DIAGNÓSTICO MÉDICO CON INCERTIDUMBRE ===\n")
sistema.agregar_hecho('fiebre', 0.9)
sistema.agregar_hecho('tos', 0.8)
sistema.agregar_hecho('dificultad_respirar', 0.6)

# Inferir
resultados = sistema.inferir()

print("\n=== RESULTADOS FINALES ===")
for hecho, certeza in sorted(resultados.items(), key=lambda x: x[1], reverse=True):
    print(f"{hecho}: {certeza:.2f} ({certeza*100:.0f}%)")
````

### Ejercicio 3: Sistema con Encadenamiento Hacia Atrás

Implementación de búsqueda de objetivos.

````python
class SistemaExpertoBackward:
    def __init__(self):
        self.reglas = []
        self.hechos = set()
        self.trace = []  # Para seguimiento del razonamiento

    def agregar_regla(self, condiciones, conclusion):
        self.reglas.append({
            'condiciones': condiciones,
            'conclusion': conclusion
        })

    def agregar_hecho(self, hecho):
        self.hechos.add(hecho)

    def probar_objetivo(self, objetivo, profundidad=0):
        """
        Intenta probar un objetivo mediante encadenamiento hacia atrás
        """
        indent = "  " * profundidad
        self.trace.append(f"{indent}¿Se puede probar '{objetivo}'?")

        # Caso base: el objetivo ya es un hecho conocido
        if objetivo in self.hechos:
            self.trace.append(f"{indent}✓ '{objetivo}' es un hecho conocido")
            return True

        # Buscar reglas que concluyan el objetivo
        for regla in self.reglas:
            if regla['conclusion'] == objetivo:
                self.trace.append(f"{indent}Encontrada regla: {regla['condiciones']} → {objetivo}")

                # Intentar probar todas las condiciones recursivamente
                todas_probadas = True
                for condicion in regla['condiciones']:
                    if not self.probar_objetivo(condicion, profundidad + 1):
                        todas_probadas = False
                        break

                if todas_probadas:
                    self.trace.append(f"{indent}✓ Todas las condiciones probadas para '{objetivo}'")
                    self.hechos.add(objetivo)  # Agregar a hechos derivados
                    return True

        self.trace.append(f"{indent}✗ No se puede probar '{objetivo}'")
        return False

    def mostrar_razonamiento(self):
        print("\n=== TRAZA DEL RAZONAMIENTO ===")
        for linea in self.trace:
            print(linea)
````

````python
# Sistema experto para elegibilidad de préstamo
sistema = SistemaExpertoBackward()

# Definir reglas
sistema.agregar_regla(['ingreso_estable', 'historial_crediticio_bueno'], 'elegible_prestamo')
sistema.agregar_regla(['empleo_formal', 'antiguedad_mayor_1año'], 'ingreso_estable')
sistema.agregar_regla(['sin_deudas_impagas', 'score_credito_alto'], 'historial_crediticio_bueno')

# Hechos conocidos del solicitante
print("=== SISTEMA DE ELEGIBILIDAD PARA PRÉSTAMO ===\n")
sistema.agregar_hecho('empleo_formal')
sistema.agregar_hecho('antiguedad_mayor_1año')
sistema.agregar_hecho('sin_deudas_impagas')
sistema.agregar_hecho('score_credito_alto')

# Intentar probar el objetivo
objetivo = 'elegible_prestamo'
print(f"Objetivo: Probar si el solicitante es '{objetivo}'\n")

resultado = sistema.probar_objetivo(objetivo)

sistema.mostrar_razonamiento()

print(f"\n=== CONCLUSIÓN ===")
print(f"¿El solicitante es elegible? {'SÍ' if resultado else 'NO'}")
````
