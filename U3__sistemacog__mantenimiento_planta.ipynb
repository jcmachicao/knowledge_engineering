# Sistema cognitivo simplilficado para mantenimiento de planta

```python
import datetime
```

```python
# --------------------------
# Memoria operativa del sistema
# --------------------------
memoria = {
    "BoilerSystem": {"last_check": datetime.date(2025, 10, 1), "freq_days": 7, "risk": "alto"},
    "PumpSystem": {"last_check": datetime.date(2025, 10, 10), "freq_days": 14, "risk": "medio"},
    "CoolingSystem": {"last_check": datetime.date(2025, 9, 30), "freq_days": 30, "risk": "bajo"}
}
```

```python
memoria
```

```python
# --------------------------
# Diccionario semántico (traducción lenguaje natural → técnico)
# --------------------------
diccionario = {
    "caldera": "BoilerSystem",
    "bomba": "PumpSystem",
    "refrigeración": "CoolingSystem",
    "presión": "PressureSensor",
    "válvula": "Valve"
}
```

```python
# --------------------------
# Rúbrica cognitiva (criterios de decisión)
# --------------------------
def evaluar_prioridad(freq_days, dias_desde_revision, riesgo):
    score = 0
    if dias_desde_revision > freq_days:
        score += 2
    if riesgo == "alto":
        score += 2
    elif riesgo == "medio":
        score += 1
    return score
```

```python
# --------------------------
# Módulo cognitivo principal
# --------------------------
def agente_mantenimiento(entrada_usuario):
    # Percepción → Traducción semántica
    for palabra, sistema in diccionario.items():
        if palabra in entrada_usuario.lower():
            target_system = sistema
            break
    else:
        return "No reconozco el sistema que mencionas."

    # Memoria → Consulta de datos
    registro = memoria.get(target_system, None)
    if not registro:
        return f"No tengo datos registrados para {target_system}."

    # Razonamiento (rúbrica)
    hoy = datetime.date.today()
    dias_desde_revision = (hoy - registro["last_check"]).days
    prioridad = evaluar_prioridad(registro["freq_days"], dias_desde_revision, registro["risk"])

    # Decisión → Recomendación
    if prioridad >= 3:
        accion = "Revisar de inmediato"
    elif prioridad == 2:
        accion = "Programar revisión esta semana"
    else:
        accion = "Condición normal"

    # Comunicación → Respuesta natural
    return (f"Sistema: {target_system}\n"
            f"Última revisión: {registro['last_check']}\n"
            f"Días desde revisión: {dias_desde_revision}\n"
            f"Nivel de riesgo: {registro['risk']}\n"
            f"Recomendación: {accion}")
```

```python
# --------------------------
# Ejemplo de interacción
# --------------------------
consulta = "¿Cómo está la presión de la caldera?"
print(agente_mantenimiento(consulta))
```
